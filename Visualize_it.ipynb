```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.barplot(data=results, x="dataset", y="train_time", hue="library")
plt.yscale("log")   # essential here — without this, sklearn's bars will be invisible
plt.ylabel("Training Time (seconds, log scale)")
plt.title("Training Time Comparison Across Libraries")
plt.tight_layout()
plt.savefig("/home/huzaifa/Desktop/my_ml_project/data/training_time_comparison.png", dpi=150)
plt.show()
```
